Example 1-6. Declaring a linear function in Python

```python
def f(x): 
    return 2 * x + 1 

x_values = [0, 1, 2, 3] 

for x in x_values: 
    y = f(x) 
    print(y)
```

Example 1-8. Charting an exponential function 


```python
from sympy import *
x = symbols('x') 
f = x **2 + 1 
plot(f) 
```

Example 1-9. Declaring a function with two independent variables in Python 

```python
from sympy import * 
from sympy.plotting import plot3d 

x, y = symbols('x y') 
f = 2 * x + 3 * y 
plot3d(f) 
```
